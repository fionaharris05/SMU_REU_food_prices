## Shorten Column Names

The original dataset contains long column names, especially for the diet-cost variables. I rename the columns used in the analysis with shorter, clearer names such as `country`, `year`, `cost_energy`, `cost_nutrient`, and `cost_healthy`. These changes are made only in the working dataset, so the original raw data remain unchanged.

```python
data_clean = data.rename(columns={
    "Classification Name": "classification",
    "Classification Code": "classification_code",
    "Country Name": "country",
    "Country Code": "country_code",
    "Time": "year",
    "Time Code": "year_code",

    "Cost of an energy sufficient diet [CoCA]": "cost_energy",
    "Cost of a nutrient adequate diet [CoNA]": "cost_nutrient",
    "Cost of a healthy diet [CoHD]": "cost_healthy",
})
```

```python
important_columns = [
    "country",
    "country_code",
    "classification",
    "year",
    "cost_energy",
    "cost_nutrient",
    "cost_healthy",
]
```

```python
diet_data = data_clean[important_columns].copy()

diet_data.head()
```

## Prepare the Diet-Cost Data for Analysis

I convert the three diet-cost columns to numeric values, treating nonnumeric entries such as `..` as missing data. Because 2017 is the only year with complete coverage across countries, I filter the dataset to that year and filter out missing costs. I updated the title of the diet categories. I then reshape the data from wide format to long format so that each row represents one country and one diet type. 

```python
cost_columns = [
    "cost_energy",
    "cost_nutrient",
    "cost_healthy"
]

diet_data[cost_columns] = diet_data[cost_columns].apply(
    pd.to_numeric,
    errors="coerce"
)
```

```python
diet_data["year"].unique()
```

```python
diet_2017 = diet_data[diet_data["year"] == 2017].copy()
```

```python
diet_long = diet_2017.melt(
    id_vars=["classification", "country", "country_code", "year"],
    value_vars=["cost_energy", "cost_nutrient", "cost_healthy"],
    var_name="diet_type",
    value_name="cost"
)
```

```python
diet_long["diet_type"] = diet_long["diet_type"].replace({
    "cost_energy": "Energy sufficient",
    "cost_nutrient": "Nutrient adequate",
    "cost_healthy": "Healthy"
})
```

```python
diet_long = diet_long.dropna(subset=["cost"])
```

```python
diet_long.head()
```

## Create Affordability Dataset

In this section, I create a separate cleaned dataset for affordability. I keep the country identifiers, year, dataset version, affordability ratios, and percentage of the population unable to afford each diet type. I then filter the data to 2017 and Food Prices for Nutrition version 2.1, convert the affordability variables to numeric values, and remove countries with missing affordability data. The final affordability dataset contains countries with complete affordability information for energy-sufficient, nutrient-adequate, and healthy diets.

```python
import pandas as pd
```

```python
raw_data = data_utf8.copy()
```

```python
raw_data.columns.tolist()
```

```python
afford_data = raw_data[
    [
        "Classification Name",
        "Country Name",
        "Country Code",
        "Time",
        "Affordability of an energy sufficient diet: ratio of cost to food expenditures [CoCA_fexp]",
        "Affordability of a nutrient adequate diet: ratio of cost to food expenditures [CoNA_fexp]",
        "Affordability of a healthy diet: ratio of cost to food expenditures [CoHD_fexp]",
        "Percent of the population who cannot afford sufficient calories [CoCA_headcount]",
        "Percent of the population who cannot afford nutrient adequacy [CoNA_headcount]",
        "Percent of the population who cannot afford a healthy diet [CoHD_headcount]"
    ]
].copy()
```

```python
afford_data = affordability_data.rename(columns={
    "Classification Name": "classification",
    "Country Name": "country",
    "Country Code": "country_code",
    "Time": "year",

    "Affordability of an energy sufficient diet: ratio of cost to food expenditures [CoCA_fexp]":
        "afford_energy",

    "Affordability of a nutrient adequate diet: ratio of cost to food expenditures [CoNA_fexp]":
        "afford_nutrient",

    "Affordability of a healthy diet: ratio of cost to food expenditures [CoHD_fexp]":
        "afford_healthy",

    "Percent of the population who cannot afford sufficient calories [CoCA_headcount]":
        "pct_cannot_afford_energy",

    "Percent of the population who cannot afford nutrient adequacy [CoNA_headcount]":
        "pct_cannot_afford_nutrient",

    "Percent of the population who cannot afford a healthy diet [CoHD_headcount]":
        "pct_cannot_afford_healthy"
})
```

```python
afford_data.columns.tolist()
```

```python
afford_keep = [
    "classification",
    "country",
    "country_code",
    "year",
    "afford_energy",
    "afford_nutrient",
    "afford_healthy",
    "pct_cannot_afford_energy",
    "pct_cannot_afford_nutrient",
    "pct_cannot_afford_healthy"
]

afford_data = afford_data[afford_keep].copy()
```

```python
afford_data = afford_data[
    (afford_data["year"] == 2017) &
    (afford_data["classification"] == "Food Prices for Nutrition 2.1")
].copy()
```

```python
afford_data.head()
```

Make sure the values are numeric

```python
afford_columns = [
    "afford_energy",
    "afford_nutrient",
    "afford_healthy",
    "pct_cannot_afford_energy",
    "pct_cannot_afford_nutrient",
    "pct_cannot_afford_healthy"
]

afford_data[afford_columns] = afford_data[afford_columns].apply(
    pd.to_numeric,
    errors="coerce"
)
```

```python
afford_data.isna().sum()
```

```python
afford_data_clean = afford_data.dropna(
    subset=afford_columns
).copy()
```

```python
afford_data_clean.isna().sum()
```
